```python
class Person:
    def __init__ (self, name, age):
        self.name = name
        self.age = age
        
    def info(self):
        return f"{self.name}, {self.age} лет"
    
    # Наследование 
class Student(Person):
    def __init__ (self, name, age):
        super().__init__(name, age)
        self.__grades = [] # Инкапсуляция
        
    def add_grade (self, grade):
            if 1 <= grade <= 5:
                self.__grades.append(grade)
            else:
                print("Оценка должна быть от 1 до 5")
        
    def get_average(self):
        if self.__grades:
            return sum(self.__grades) / len(self.__grades)
        return 0
    # Полиморфизм
    def info(self):
        return f"Студент: {self.name}, {self.age} лет, Средняя оценка: {self.get_average():.2f}"
    
class Teacher(Person):
    def __init__(self, name, age, subject):
        super().__init__(name, age)
        self.subject = subject

    def info(self):
        return f"Преподаватель: {self.name}, предмет: {self.subject}"

# 📌 Работа программы
people = [
    Student("Анна", 20),
    Student("Иван", 22),
    Teacher("Петров", 45, "Математика")
]

# добавим оценки
people[0].add_grade(5)
people[0].add_grade(4)

people[1].add_grade(3)
people[1].add_grade(4)

# полиморфизм в действии
for person in people:
    print(person.info())
```
