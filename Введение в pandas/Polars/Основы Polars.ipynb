```python
import polars as pl
```

```python
data = {
    "имя": ["Анна", "Борис", "Виктор", "Анна", "Дарья"],
    "товар": ["Кофе", "Чай", "Печенье", "Чай", "Кофе"],
    "количество": [2, 1, 3, 2, 1],
    "цена_за_ед": [350, 120, 80, 120, 350],
    "город": ["Мск", "Спб", "Мск", "Мск", "Казань"]
}

df = pl.DataFrame(data)
```

```python
df_select = df.select(pl.col('имя'), pl.col('город'))
df_select1 = df.select(["имя", "город"])
df_moscow = df.filter(pl.col('город') == 'Мск')
df_with_total = df.with_columns(
    (pl.col('количество') * pl.col('цена_за_ед')).alias('итого')
)

lazy_plan = (
    df.lazy()
    .filter(pl.col("город") == "Мск")
    .with_columns(
        (pl.col("количество") * pl.col("цена_за_ед")).alias("итого")
    )
    .group_by('товар')
    .agg(
        pl.col('количество').sum().alias('всего_продано'),
        pl.col('итого').sum().alias('общая_выручка')
)
    .sort('общая_выручка',descending=True)
)

result = lazy_plan.collect()
result
```
